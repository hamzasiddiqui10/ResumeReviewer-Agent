# Environment & API Key Verification
Concept: Environment variables store secret keys safely outside source code. We use python-dotenv to populate os.environ from ./.env, then perform quick zero-token smoke tests against Tavily and Gemini to confirm API authentication before building abstractions.

```python
import os
from dotenv import load_dotenv
from tavily import TavilyClient
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq

# 1. Load environment variables from .env
load_dotenv()

# Verify presence of required keys
required_keys = ["GEMINI_API_KEY", "GROQ_API_KEY", "TAVILY_API_KEY"]
for key in required_keys:
    status = "Loaded" if os.getenv(key) else "MISSING"
    print(f"{key}: {status}")
```

```python
# 2. Smoke test Tavily Search API
tavily_client = TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))
search_test = tavily_client.search(
    query="Machine Learning Engineer Germany site:linkedin.com/jobs", 
    max_results=1
)
print(f"Tavily API Status: Connected ({len(search_test.get('results', []))} test result returned)")
```

```python
# 3. Smoke test Gemini via LangChain
llm_test = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash", 
    google_api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
prompt = "Respond with 'LLM Connected' if working."
response = llm_test.invoke(prompt)
print(f"Gemini API Status: {response.text}")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash", 
    google_api_key=os.getenv("GEMINI_API_KEY")
)

prompt = [
    ["system", "You are a smart geography teacher who gives to the point answers to students' questions."],
    ["human", "How many provinces are there in Pakistan?"],
]
```

```python
response = llm.invoke(prompt)
print(response.text)
print("-"*50)
print(response.content)
print("-"*50)
```

```python
# Groq API test
llm = ChatGroq(
    model="qwen/qwen3-27b", 
    api_key=os.getenv("GROQ_API_KEY"),
    reasoning_format="parsed",
)

prompt = [
    ["system", "You are a smart geography teacher who gives to the point answers to students' questions."],
    ["human", "What are top 5 richest cities in Germany?"],
]
```
